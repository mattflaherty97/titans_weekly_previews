# Week 6 vs Indianapolis

I hope everyone is well-rested from the bye week in week 5. A bye week that was much needed as Jeffrey Simmons was a late scratch from week 4, and Levis hurt his shoulder during the game. Now, we will be settling in for 13 straight weeks of Titans' football, and what better way to kick this stretch of games off than with a divisional matchup against the Colts?

# Overall Preview

While the Titans scored over 30 points for the first time since January 2, 2022, the offense was saved by the defense which contributed 2 points via safety. Nevertheless, a 29 point performance from this iteration of the offense was remarkable considering a majority of the game was played with a backup quarterback that generated an explosive pass rate which ranked 29th in week 4. Sacks still remain the thorn in the Titans passing game as they ranked 20th in EPA/sack in week 4 leaving them at 30th on the season. While the explosive plays are not there, the Titans will need to figure out a way to decrease the sacks in order to rejuvenate this passing game. Fortunately, the Colts blitz at the second lowest rate and only rank 19th in pressure rate. This should allow Levis (should he be healthy to play) to settle into the game and work through his reads without being rushed. Levis must take advantage of this because he ranks second worst in sacks that are the quarterback's fault as charted by FTN. While the passing game is still working through some things, the rushing attack was able to take over as it was the 10th most efficient rushing attack in week 4; however, the Colts have the 10th best rush defense coming into week 6.

For the Titans defense, they finally generated their first turnover of the year on a backwards pass (charted as a run) from Tyler Huntley. The pass defense has been superb despite any interceptions this year ranking 10th in EPA/pass and 9th in explosive plays allowed per pass. They have managed to do this despite a 24th ranked pressure rate and a blitz rate that is ranked 25th. I think the Titans should look to increase the pressure in this game against Joe Flacco who has the 4th quickest pocket time and the 4th lowest pressure rate. Even if Anthony Richardson is able to play, this strategy should still be successful as Richardson has a long pocket time and is susceptible to pressure. Will the Colts look to run the ball against such a successfull pass defense? Week 5 stats (Jonathan Taylor out) don't suggest they'll lean on the run. In week 5, the Colts had the 5th highest pass rate over expected. Even if Taylor is cleared to play, the Colts will still have difficulty running as they ranked 24th in EPA/rush in weeks 1-4 with Taylor. 

# Trenches

The Titans continue to have the most success running behind JC Latham; however, the Colts right side of the defensive line is their strongest side. This should be a good matchup to watch. The Titans are using the 10th highest motion rate on run plays and I think they should continue to use motion to try to take some pressure off the offensive line. They have also benefitted from using motion as they rank 16th in EPA/rush with motion while ranking 20th without motion.

The Colt are also seeing more success running behind the left side of the offensive line. They might need to utilize outside runs to the left side of the offensive line to avoid T'Vondre Sweat who has helped lead the way on this Titans strong run defense. 

# Passing Heatmap

The Titans passing attack has yet to get going this season. Some of the struggles are caused by the defenses the Titans have played thus far: Chicago, New York Jets, Green Bay, and Miami. Three of the four rank in the top 10 while all four rank in the top 15. However, the Colts rank 25th in EPA/pass. As with the run, using motion might help get the defense off balance and free up some areas on the field in which the Titans have struggled to pass the ball. I will go one step further and say that the Titans should reduce play action and try to keep Levis in the pocket with true dropbacks. Coach Callahan should get him to keep his eyes down field and get him some quick reads.

As mentioned above, the Colts are stronger passers than runners. The Titans should have success if they can force the ball to the perimeter and heat up Flacco/Richardson. Both options at quarterback have decent deep balls; therefore, bringing pressure and not allowing the quarterback the time to wait on his receiver to get open could help the Titans' defense reduce explosive plays.

# Pass Catchers

Ridley continues to lead the way in targets. Unfortunately, while Hopkins has led the way in targets for the last two games, he is still the least efficient pass catcher for the Titans. Josh Whyle is not a household name (yet?), but I would be interested to see how his prescence in the passing game affects the efficiency. While only being targeted 8 times, he has been the most efficient pass catcher for the offense while having a similar number of targets as fellow tight end Chig Okonkwo.

As for the Colt pass catchers, the deep threat of Alec Pierce leads the way in efficiency while Michael Pittman has had the majority of the target share. Pittman will miss week 6 with an injury which will allow Josh Downs to see more targets and maybe we finally see a breakout game from the rookie, AD Mitchell. This is a decent pass catching group; however, how will the abscence of Pittman and the potential abscence of Downs affect the passing game?

# Rushers

Pollard continues to lead the way in rushes for the Titans and is even beginning to take a majority of the snaps as he has ecplised 62% of the snaps. He has been the most efficient rusher on the team for ball carries that have a substantial number of carries. However, Levis, with a smaller sample size, has been quite an efficient rusher on his 13 carries. While exposing your potential franchise quarterback to hits is not always the best thing to do, Levis adds values to the run game.

The Colts could be expected to miss Jonathan Taylor for a second week in a row with an injury. Week 5 saw Trey Sermon get the most carries with 10 followed by Tyler Goodson with 5. It will be interesting to see how the Colts backfield is handled this week as Goodson was able to be more efficient with his rushes and receptions than Sermon in a limited role.

# Prediction

Predictions are 3-2 after week 4 at Miami with a crazy backdoor cover for the over. This week, I think the Titans will cover -2.5, and the game will go under 43.5. The Titans offense hasn't been great this year; however, defensively, the Colts haven't been much better. Levis isn't playing a strong pass rush, so this should allow him to reduce his negative plays that have hindered his performance thus far. I think the Titans' defense will be the determining factor in the outcome. Can they pressure Joe Flacco/Anthony Richardson and finally get their first interception of the season? The odds are favoring this event as the injuries pile of for the Colts' offense. I think this game will go under 43.5 because it's not going to be pretty when the Titans have the ball, and I think the Titans' defense should be able to handle this offense that could be missing 58% of its carries and 26% of its targets.
